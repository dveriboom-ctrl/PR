## 1) Что такое PR и зачем он нужен

PR — это предложение влить изменения из вашей ветки в основную ветку (обычно `main`). Это:
- Контроль качества (ревью, автотесты, линтеры);
- История обсуждений и решений;
- Прозрачный процесс: кто, зачем и что меняет.

Правила хорошего PR:
- Делайте PR **маленькими** (легче ревьюить и откатывать);
- Пишите **чёткое описание**: что изменили, почему, как тестировали;
- Прикладывайте скриншоты/видео для UI‑изменений;
- Добавляйте **чек‑лист** (миграции? тесты? дока?).

## 2) Минимальный Git‑флоу

Ниже — базовые шаги. Выполните их в своём репозитории (в терминале), не в ноутбуке.

````python
# 2.1 Клонирование и настройка апстрима (если у вас форк)
git clone https://github.com/USER/REPO.git
cd REPO
git remote add upstream https://github.com/ORIGINAL_OWNER/REPO.git  # опционально

# 2.2 Создание фиче-ветки от main
git checkout -b feat/readme-typo-fix

# 2.3 Коммит изменений
git add README.md
git commit -m "docs: fix typos in README"

# 2.4 Push ветки
git push -u origin feat/readme-typo-fix

# 2.5 Создание PR
# Откройте GitHub в браузере: появится кнопка "Compare & pull request"
````

## 6) Практика: смержить PR правильно

- Дождитесь зелёных проверок CI.
- Исправьте замечания ревьюера / Gemini.
- Выберите стратегию: **Squash & merge** (чаще всего), **Rebase & merge** или **Merge commit**.
- После мерджа **удалите фиче‑ветку** и **обновите локальный `main`**.

````python
# Обновление локального main после мерджа
git checkout main
git pull --ff-only
git branch -d feat/readme-typo-fix  # удаляем локальную ветку
git push origin --delete feat/readme-typo-fix  # удаляем удалённую ветку (опционально)
````

## 7) Типичные проблемы и как их решать

- **Конфликты:** обновите `main`, ребейз/мердж в вашу ветку, решите конфликты, повторите push.
- **Красный CI:** смотрите логи в вкладке *Actions* → конкретный job.
- **Слишком большой PR:** разбейте на несколько небольших, с ясными целями.
